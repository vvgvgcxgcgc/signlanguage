```python
import os
import subprocess
from pathlib import Path

candidates = (
    Path("/kaggle/working/signlanguage"),
    Path.cwd(),
    Path.cwd().parent,
)
root = next((path for path in candidates if (path / "training" / "runs").is_dir()), None)
if root is None:
    raise FileNotFoundError("training/runs was not found from the notebook working directory")

os.chdir(root)
scripts = sorted((root / "training" / "runs").glob("*.sh"))
if not scripts:
    raise FileNotFoundError(f"no bash scripts in {root / 'training' / 'runs'}")

for script in scripts:
    print(f"=== {script.name} ===", flush=True)
    completed = subprocess.run(["bash", str(script)], cwd=root)
    subprocess.run(["pkill", "-f", "python -m training.run"], check=False)
    if completed.returncode != 0:
        raise SystemExit(completed.returncode)
```
